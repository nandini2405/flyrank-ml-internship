The starter dataset contains 30,000 pages. Of these, 16,262 pages have a declining trend, and 9,961 pages have both a declining trend and at least 500 impressions over the last 90 days.

These numbers show that the dataset contains a substantial set of pages with measurable search visibility and evidence of decline, making it useful to investigate how page-level signals can help prioritize pages for review.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/nandini2405/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Total pages:", len(df))

print("Declining pages:", (df["trend_direction"] == "down").sum())

print("Declining pages with 500+ impressions:",
      ((df["trend_direction"] == "down") &
       (df["impressions_90d"] >= 500)).sum())
```
