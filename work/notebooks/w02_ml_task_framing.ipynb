The proxy label is based on the observed `trend_direction` field, where `down` represents a declining page.

This is a defined label from the current data window rather than a future observed outcome. Therefore, it is only a proxy for review priority and does not mean that a page will recover after a refresh.

## 3. Success metric

*One metric you can defend. What number means 'good'?*


I will use **Precision@20** as my success metric.

Precision@20 measures how many of the top 20 pages ranked for review are relevant according to the target or proxy.

I would consider a Precision@20 of **0.70 or higher** a useful result, meaning at least 14 of the top 20 ranked pages match the defined target.

This metric fits the problem because a reviewer has limited capacity and will review the highest-ranked pages first.
